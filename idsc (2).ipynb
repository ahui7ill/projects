```python
#cleaning function
def clean(df, name, type):
    #drop unwanted columns
    if len(df.columns) == 7:
        df = df.drop(columns=['Unnamed: 2','Unnamed: 5'])
    else:
        df = df.drop(columns=['Unnamed: 2','Unnamed: 5', 'Unnamed: 7', 'Unnamed: 8'])

    #rename columns
    df.columns = ['Form', 'Average Retail Price (per pound)', 'Preparation Yield Factor', 'Size of a cup equivalent', 'Average Price per cup equivalent']
    
    #add fruit column with fruit name and move to front
    if type == 'fruit':
        df['Fruit'] = name.capitalize()
        df = df[ ['Fruit'] + [ col for col in df.columns if col != 'Fruit' ] ]
    elif type == 'veg':
        df['Vegetable'] = name.capitalize()
        df = df[ ['Vegetable'] + [ col for col in df.columns if col != 'Vegetable' ] ]

    #only keep rows with Fresh1
    df = df[(df.Form == 'Fresh1') | (df.Form == 'Fresh1 ')]
   
    #change Fresh1 to Fresh
    df.replace({'Form':{'Fresh1':'Fresh','Fresh1 ':'Fresh'}}, inplace=True)
    return df
```

```python
p = Path('fruit_2016')
fruit_2016 = pd.DataFrame()
# Iterate through files in Path object, clean, and concatenate to fruit dataframe
for p in p.iterdir():
    try: 
        df = pd.read_csv(p, encoding='unicode_escape')
        df = clean(df, p.stem.split(' ')[0], 'fruit')
        fruit_2016 = pd.concat([fruit_2016, df], ignore_index=True).drop_duplicates()
        fruit_2016['Year'] = '2016'
        fruit_2016['Average Retail Price (per pound)'] = fruit_2016['Average Retail Price (per pound)'].replace({'\$':''}, regex = True)

    except Exception as e:
        continue

# 4 missing rows: cranberries, dates, figs, and fruit_cocktail because they are not fresh
fruit_2016
```

```python
import numpy as np

def load_glove_model(File):
    print("Loading Glove Model")
    glove_model = {}
    with open(File,'r', encoding='utf-8') as f:
        for line in f:
            split_line = line.split()
            word = split_line[0]
            embedding = np.array(split_line[1:], dtype=np.float64)
            glove_model[word] = embedding
    print(f"{len(glove_model)} words loaded!")
    return glove_model
```

```python
glove_model = load_glove_model('glove.42B.300d.txt')
```

```python
#visualize pca of word embeddings

from sklearn.decomposition import PCA

import matplotlib.pyplot as plt

# model = gensim.downloader.load("glove-wiki-gigaword-300")

# twitter_model= gensim.downloader.load("glove-twitter-100")

X=[glove_model[word] for word in vegs]

#identify clusters of words in the word embeddings at high dimensions

pca=PCA(n_components=2)

X_pca=pca.fit_transform(X)

plt.scatter(X_pca[:,0],X_pca[:,1])

for i, word in enumerate(vegs):
    plt.annotate(word, (X_pca[i,0],X_pca[i,1]))

plt.show()
```

```python
Y=[glove_model[word] for word in fruit]

pca=PCA(n_components=2)

Y_pca=pca.fit_transform(Y)

plt.scatter(Y_pca[:,0],Y_pca[:,1])

for i, word in enumerate(fruit):
    plt.annotate(word, (Y_pca[i,0],Y_pca[i,1]))

plt.show()
```
